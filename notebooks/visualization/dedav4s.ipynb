# # Plotting with [dedav4s](https://quafadas.github.io/dedav4s)
#
# [dedav4s](https://github.com/Quafadas/dedav4s) (declarative data visualization for Scala) is a Scala 3 library
# built around [Vega](https://vega.github.io/vega) and [Vega-Lite](https://vega.github.io/vega-lite) specifications:

#
# Its `dedav4s_notebook` module shows plots in almond. Each plot comes with an interactive Vega view, that JupyterLab,
# Jupyter Notebook and VS Code render, and a static SVG, that viewers which don't run JavaScript (like GitHub) show instead.
#

# ## Quick plots
#
# dedav4s adds methods to plot Scala collections directly. These fill the data of one of the specs that ship with
# dedav4s, and accept functions that modify the spec further, here to change its title:

```scala
import viz.extensions.*

val rng = new scala.util.Random(42)

val points = Seq.fill(100) {
  val x = rng.nextDouble() * 10
  x -> (2 * x + 3 + rng.nextGaussian() * 2)
}

points.plotScatter(List(spec => spec("title") = "Random points"))
```

```scala
points.plotRegression(List(spec => spec("title") = "Linear regression"))
```

```scala
Seq(3, 1, 4, 1, 5, 9, 2, 6, 5, 3, 5).plotLineChart(List(spec => spec("title") = "Digits of pi"))
```

# ## Starting from a spec
#
# `VegaPlot.fromString` parses a Vega or Vega-Lite spec at compile time. Its `plot` method then accepts changes to any of
# the fields of the spec, like its data or title, and it shows the modified spec. Changing a field that the
# spec doesn't have, or setting a field to a value of the wrong type, doesn't compile.

```scala
val barChart = VegaPlot.fromString("""{
  "$schema": "https://vega.github.io/schema/vega-lite/v5.json",
  "title": "A bar chart",
  "data": {"values": [{"category": "A", "amount": 28}]},
  "mark": "bar",
  "encoding": {
    "x": {"field": "category", "type": "nominal", "axis": {"labelAngle": 0}},
    "y": {"field": "amount", "type": "quantitative"}
  }
}""")

barChart.plot(
  _.title := "Lines of code per language",
  _.data.values := Seq(
    (category = "Scala", amount = 12000),
    (category = "Java", amount = 8000),
    (category = "Kotlin", amount = 1500),
    (category = "Python", amount = 3000)
  )
)
```

# Changes can also merge JSON into the spec, here to change the mark and to add a field to the encoding:

```scala
val scatterPlot = VegaPlot.fromString("""{
  "$schema": "https://vega.github.io/schema/vega-lite/v5.json",
  "title": "Clusters",
  "width": 400,
  "height": 300,
  "data": {"values": []},
  "mark": "point",
  "encoding": {
    "x": {"field": "x", "type": "quantitative"},
    "y": {"field": "y", "type": "quantitative"}
  }
}""")

locally {
  val clusters = for {
    (center, cluster) <- Seq((-2.0, -1.0), (0.0, 2.0), (2.5, -0.5)).zipWithIndex
    _ <- 1 to 100
  } yield (
    x = center._1 + rng.nextGaussian() * 0.6,
    y = center._2 + rng.nextGaussian() * 0.6,
    cluster = s"cluster ${cluster + 1}"
  )

  scatterPlot.plot(
    _.data.values := clusters,
    _.mark := (`type` = "circle", size = 40).asJson,
    _.encoding += (color = (field = "cluster", `type` = "nominal")).asJson
  )
}
```

# ## Transforms
#
# Vega and Vega-Lite can transform the data themselves (binning, aggregating, filtering, …), so that we can pass them raw values.
# Here, Vega-Lite bins and counts values to draw a histogram:

```scala
val samples = Seq.fill(1000)(rng.nextGaussian()) ++ Seq.fill(500)(3 + 0.7 * rng.nextGaussian())

val histogram = VegaPlot.fromString("""{
  "$schema": "https://vega.github.io/schema/vega-lite/v5.json",
  "title": "Histogram",
  "width": 400,
  "height": 250,
  "data": {"values": []},
  "mark": "bar",
  "encoding": {
    "x": {"field": "value", "type": "quantitative", "bin": {"maxbins": 40}},
    "y": {"aggregate": "count", "type": "quantitative"}
  }
}""")

histogram.plot(
  _.data.values := samples.map(v => (value = v))
)
```

# ## ECharts
