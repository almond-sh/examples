# sttp
A minimal setup of [sttp](https://sttp.softwaremill.com/)

```scala
import $ivy.`com.softwaremill.sttp.client4::core:4.1.0`
import sttp.client4.quick._
quickRequest.get(uri"https://httpbin.org/uuid").send()
```

## Async sttp
Use the `Future` backend of sttp, which sends requests asynchronously.

```scala
import sttp.client4.DefaultFutureBackend
import scala.concurrent.ExecutionContext.Implicits.global

val backend = DefaultFutureBackend()

val r = basicRequest.get(uri"https://httpbin.org/uuid")
```

```scala
import scala.concurrent.Await
import scala.concurrent.duration._

val future = r.send(backend)
Await.result(future, 1.minute).body
```

# Armeria
The client of [Armeria](https://armeria.dev/), a Java library

```scala
import $ivy.`com.linecorp.armeria:armeria:1.42.0`

import com.linecorp.armeria.client.WebClient

val client = WebClient.of("https://httpbin.org/")
val future = client.get("/uuid").aggregate()
val res = future.join()
```

```scala
val body = res.contentUtf8()
```
